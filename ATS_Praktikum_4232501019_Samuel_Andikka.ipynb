```python
#Soal 5.Katup Pengaman (20 poin | CLO-1 | TP-3, TP-4)

#Membaca tekanan uap dari input
tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

#Status manual override
status_manual_override = False

#Menentukan status katup
if tekanan_uap > 150 and status_manual_override == False:
    print("Katup Terbuka")

elif status_manual_override == True:
    print("Katup Mode Manual")

else:
    print("Katup Tertutup")
```

```python
#Soal 7.Pemanas Mencapai Target (25 poin | CLO-1 | TP-5)

#Suhu awal
suhu_saat_ini = 60

#Perulangan while sampai mencapai target 85 C
while suhu_saat_ini < 85:
    suhu_saat_ini += 5

    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

#pesan setelah target tercapai
print("Target 85 C tercapai.")
```

```python
#Soal 9.Pembacaan Vibrasi dengan Sensor Rusak (30 poin | CLO-1 | TP-5, TP-6)

#Data vibrasi lima unit mesin
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

#Penghitung data tidak valid
jumlah_tidak_valid = 0

#Memeriksa setiap data
for i, nilai in enumerate(data, start=1):

    try:
        vibrasi = float(nilai)

    except ValueError:
        print(f"Unit {i}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    #Memeriksa vibrasi tinggi
    if vibrasi > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")

    #Memeriksa kondisi kritis
    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

#Menampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

D. Try dan Except Try digunakan untuk mengubah data menjadi angka desimal. Except ValueError menangani data yang tidak bisa diubah menjadi angka. Jumlah data tidak valid akan bertambah satu.

E. Continue Continue digunakan untuk melewati data yang tidak valid dan melanjutkan pemeriksaan ke data berikutnya.

F. Pemeriksaan Vibrasi Tinggi If digunakan untuk memeriksa vibrasi di atas 2.0. Jika terpenuhi, program menampilkan peringatan.

G. Kondisi Kritis Jika vibrasi lebih dari 3.0, program menampilkan pesan kritis. Break digunakan untuk menghentikan perulangan.

H. Hasil Akhir Fungsi print() menampilkan jumlah data yang tidak valid. Pada data tersebut terdapat 1 data tidak valid, yaitu ERR.

```python
#Challenge: Bonus Nilai (hingga +25 poin) (CLO-1 | TP-4, TP-5, TP-6)

#PROGRAM PEMANTAUAN KONSUMSI LISTRIK HARIAN
#Program ini digunakan untuk memantau konsumsi
#listrik selama 7 hari dan menentukan kategori
#penggunaan listrik berdasarkan total konsumsi.

print("  PROGRAM PEMANTAUAN KONSUMSI LISTRIK")


#List untuk menyimpan data konsumsi listrik
konsumsi_listrik = []

#Memasukkan data konsumsi listrik selama 7 hari
for hari in range(1, 8):
    try:
        energi = float(input(f"Masukkan konsumsi listrik hari ke-{hari} (kWh): "))

        if energi < 0:
            print("Konsumsi tidak boleh negatif!")
            energi = 0

        konsumsi_listrik.append(energi)

    except ValueError:
        print("Input tidak valid! Data diisi 0 kWh.")
        konsumsi_listrik.append(0)

#Menghitung total konsumsi listrik
total = 0

for energi in konsumsi_listrik:
    total += energi

#Menghitung rata-rata konsumsi listrik
rata_rata = total / len(konsumsi_listrik)

#Menentukan kategori konsumsi listrik
if total > 100:
    kategori = "Konsumsi listrik tinggi"
elif total >= 50:
    kategori = "Konsumsi listrik sedang"
else:
    kategori = "Konsumsi listrik rendah"

#222Menampilkan hasil pemantauan
print("       HASIL PEMANTAUAN LISTRIK")

for i, energi in enumerate(konsumsi_listrik, start=1):
    print(f"Hari ke-{i}: {energi:.2f} kWh")

print(f"Total konsumsi   : {total:.2f} kWh")
print(f"Rata-rata harian : {rata_rata:.2f} kWh")
print(f"Kategori         : {kategori}")
print("Pemantauan selesai.")
```

PENJELASAN PROGRAM PEMANTAUAN LISTRIK

A. List Konsumsi Listrik List digunakan untuk menyimpan data konsumsi listrik selama 7 hari.

B. Perulangan For For digunakan untuk meminta input konsumsi listrik setiap hari. Range(1, 8) digunakan untuk mengulang sebanyak 7 kali.

C. Try dan Except Try digunakan untuk membaca input berupa angka desimal. Except ValueError menangani input yang tidak valid. Jika input salah, konsumsi otomatis diisi 0 kWh.

D. Percabangan If If digunakan untuk memeriksa apakah konsumsi bernilai negatif. Jika negatif, nilai diubah menjadi 0 kWh.

E. Append Append digunakan untuk memasukkan data konsumsi ke dalam list.

F. Menghitung Total dan Rata-rata Perulangan for menjumlahkan seluruh konsumsi listrik. Rata-rata dihitung dengan membagi total dengan jumlah data. Fungsi len() digunakan untuk menghitung jumlah data.

G. Menentukan Kategori If, elif, dan else digunakan untuk menentukan kategori konsumsi. Di atas 100 kWh = konsumsi tinggi. 50 sampai 100 kWh = konsumsi sedang. Di bawah 50 kWh = konsumsi rendah.

H. Menampilkan Hasil Enumerate digunakan untuk memberikan nomor pada setiap hari. Format .2f digunakan untuk menampilkan dua angka desimal. Print() digunakan untuk menampilkan total, rata-rata, dan kategori.
