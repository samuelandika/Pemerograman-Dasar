```python
# AKTIVITAS 6F

scores = [80, 65, 90, 75, 70]

total = 0                    # accumulator: starts at zero
count_pass = 0               # a counter is an accumulator that adds 1
for score in scores:
    total = total + score    # add each element
    if score >= 75:
        count_pass = count_pass + 1
    print("running total:", total)

average = total / len(scores)
print("Total   :", total)
print("Average :", average)
print("Scores >= 75:", count_pass, "of", len(scores))
```

```python
# AKTIVITAS 6G

temps = [28, 31, 30, 33, 29, 27, 32]

# 5-day moving average: window = the 5 values ending at index i
avg5 = []
for i in range(4, len(temps)):
    window = temps[i - 4 : i + 1]
    avg5.append(round(sum(window) / 5, 1))

print("5-day averages:", avg5)
print(len(temps), "values ->", len(avg5), "averages")
```

```python
# AKTIVITAS 6H

# Data produksi energi 30 hari (1 - 30 November 2026) dalam MWh
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083, # Hari 1-10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098, # Hari 11-20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062  # Hari 21-30
]

# Array paralel untuk nomor hari, jam operasi, dan nama hari
days = list(range(1, 31))
run_hours = [24] * 30
run_hours[16] = 18  # Hari ke-17 (indeks 16) hanya beroperasi 18 jam
DAY_NAMES = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]

print("=== 1. DETEKSI PENURUNAN ENERGI (> 20%) ===")
for i in range(1, len(energy)):
    perubahan = (energy[i] - energy[i - 1]) / energy[i - 1] * 100
    if perubahan < -20:
        print(f"Hari ke-{days[i]} ({DAY_NAMES[i % 7]}): Turun {perubahan:.1f}% dari hari sebelumnya")
        print(f"  -> Energi hari ke-{days[i-1]}: {energy[i-1]} MWh")
        print(f"  -> Energi hari ke-{days[i]}: {energy[i]} MWh")

print("\n=== 2. CEK JAM OPERASI TIDAK PENUH (< 24 JAM) ===")
for i in range(len(energy)):
    if run_hours[i] < 24:
        selisih_jam = 24 - run_hours[i]
        print(f"Hari ke-{days[i]} ({DAY_NAMES[i % 7]}): Beroperasi {run_hours[i]} jam (mati selama {selisih_jam} jam)")

print("\n=== 3. PERBANDINGAN RATA-RATA HARI KERJA VS AKHIR PEKAN ===")
hari_kerja = []
akhir_pekan = []

for i in range(len(energy)):
    if DAY_NAMES[i % 7] == "Sab" or DAY_NAMES[i % 7] == "Min":
        akhir_pekan.append(energy[i])
    else:
        hari_kerja.append(energy[i])

print(f"Rata-rata Hari Kerja  : {sum(hari_kerja) / len(hari_kerja):.0f} MWh")
print(f"Rata-rata Akhir Pekan : {sum(akhir_pekan) / len(akhir_pekan):.0f} MWh")
```

```python
# AKTIVITAS 6I

# Membuat daftar temps dengan 7 suhu harian (°C)
temps = [28, 31, 30, 33, 29, 27, 32]

# (1) Tampilkan 3 hari pertama dan 2 hari terakhir dengan memotong (slicing)
print("3 hari pertama :", temps[:3])
print("2 hari terakhir:", temps[-2:])

# (2) Tambahkan satu hari dengan append, lalu keluarkan lagi dengan pop
temps.append(34)
print("Setelah append :", temps)

suhu_dihapus = temps.pop()
print("Setelah pop    :", temps, "(dihapus:", suhu_dihapus, ")")

# (3) Tampilkan versi yang diurutkan dengan sorted tanpa mengubah temps
print("Terurut        :", sorted(temps))
print("Asli (utuh)    :", temps)

# (4) Hitung rata-rata dan temukan suhu tertinggi dan hari apa itu
rata_rata = sum(temps) / len(temps)
suhu_tertinggi = max(temps)
hari_tertinggi = temps.index(suhu_tertinggi) + 1  # + 1 karena indeks dimulai dari 0

print("Rata-rata      :", round(rata_rata, 1), "°C")
print("Suhu tertinggi :", suhu_tertinggi, "°C pada hari ke-", hari_tertinggi)

# (5) Buat daftar baru suhu di atas rata-rata
di_atas_rata_rata = []
for t in temps:
    if t > rata_rata:
        di_atas_rata_rata.append(t)

print("Di atas rata2  :", di_atas_rata_rata)
```

```python
# PEMANASAN

scores = [80, 65, 90, 75, 70, 85]

# (1) Mencetak skor pertama dan skor terakhir (dengan indeks negatif)
print("1. Skor pertama:", scores[0], "| Skor terakhir:", scores[-1])

# (2) Mencetak rata-rata
rata_rata = sum(scores) / len(scores)
print("2. Rata-rata:", rata_rata)

# (3) Menghitung berapa banyak skor yang berada di atas rata-rata
di_atas_rata = 0
for s in scores:
    if s > rata_rata:
        di_atas_rata += 1
print("3. Jumlah skor di atas rata-rata:", di_atas_rata)

# (4) Mencetak posisi skor terendah sebagai nomor siswa (siswa pertama = 1)
pos_min = 0
for i in range(len(scores)):
    if scores[i] < scores[pos_min]:
        pos_min = i
print("4. Posisi skor terendah: Siswa ke-", pos_min + 1)

# (5) Membuat daftar baru dengan setiap skor ditambah 5
skor_baru = []
for s in scores:
    skor_baru.append(s + 5)
print("5. Daftar skor baru (+5):", skor_baru)
```

```python
# LATIHAN LABORATORIUM

# Data produksi energi harian selama 30 hari (1-30 November 2026) dalam MWh
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,  # hari 1–10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,  # hari 11–20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062   # hari 21–30
]

# Input target produksi harian dari pengguna
target_input = input("Masukkan target energi harian (MWh) [contoh: 5800]: ")
target = float(target_input) if target_input.strip() else 5800.0

print("\n" + "=" * 52)
print("       LAPORAN PRODUKSI ENERGI BULANAN")
print("=" * 52)

# 1. Total dan rata-rata energi bulanan
total_energy = sum(energy)
avg_energy = total_energy / len(energy)

print(f"Total Energi Bulanan : {total_energy:,} MWh")
print(f"Rata-rata Harian     : {avg_energy:,.2f} MWh/hari")

# 2. Tiga hari tertinggi dan tiga hari terendah (nomor hari dan nilainya)
# Membuat daftar pasangan (nilai_energi, nomor_hari) agar mudah diurutkan
indexed_energy = []
for i in range(len(energy)):
    indexed_energy.append((energy[i], i + 1))

# Urutkan dari terendah ke tertinggi tanpa merusak list asli
sorted_energy = sorted(indexed_energy)

lowest_3 = sorted_energy[:3]
highest_3 = sorted_energy[-3:][::-1]  # Ambil 3 terakhir lalu balik urutannya

print("\n--- 3 Hari Produksi Tertinggi ---")
for rank, (val, day) in enumerate(highest_3, start=1):
    print(f"Peringkat {rank} : Hari ke-{day:<2} -> {val:,} MWh")

print("\n--- 3 Hari Produksi Terendah ---")
for rank, (val, day) in enumerate(lowest_3, start=1):
    print(f"Peringkat {rank} : Hari ke-{day:<2} -> {val:,} MWh")

# 3. Jumlah hari di bawah target yang dimasukkan
below_target_count = 0
for val in energy:
    if val < target:
        below_target_count += 1

print(f"\n--- Evaluasi Target ({target:,.0f} MWh) ---")
print(f"Jumlah hari di bawah target : {below_target_count} hari dari {len(energy)} hari")

# 4. Hari yang rata-rata pergerakan 7 harinya turun di bawah 97% dari rata-rata bulanan
threshold_97 = 0.97 * avg_energy
print(f"\n--- Peringatan Rata-rata Pergerakan 7 Hari (< 97% atau < {threshold_97:,.2f} MWh) ---")

found_warning = False
for i in range(6, len(energy)):
    window_7 = energy[i - 6 : i + 1]
    moving_avg_7 = sum(window_7) / 7
    if moving_avg_7 < threshold_97:
        print(f"Hari ke-{i + 1:<2} : Rata-rata 7 hari = {moving_avg_7:,.2f} MWh")
        found_warning = True

if not found_warning:
    print("Tidak ada hari dengan rata-rata 7 hari di bawah batas 97%.")

# 5. Total energi per minggu (Minggu 1–4, ditambah hari-hari yang tersisa)
print("\n--- Total Energi Per Minggu ---")
for week in range(4):
    start_idx = week * 7
    end_idx = start_idx + 7
    week_total = sum(energy[start_idx:end_idx])
    print(f"Minggu ke-{week + 1} (Hari {start_idx + 1:02d}-{end_idx:02d}) : {week_total:,} MWh")

# Hari-hari yang tersisa (Hari ke-29 dan ke-30)
remaining_days = energy[28:]
remaining_total = sum(remaining_days)
print(f"Sisa Hari   (Hari 29-30) : {remaining_total:,} MWh")
print("=" * 52)
```
